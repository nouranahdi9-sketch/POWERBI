## dim_processes_phases




```python
dim_processes_phases = processes_phases.filter(F.col("deleted") == False)

# Le rapprochement depuis gold.phases se fait sur `code` : il doit etre unique.
doublons_code = (
    dim_processes_phases
    .groupBy("code").count().filter(F.col("count") > 1)
)
if doublons_code.count() > 0:
    display(doublons_code)
    raise ValueError(
        "processes_phases : `code` n'est pas unique parmi les lignes actives. "
        "Le rapprochement depuis gold.phases[prd_workshop] serait ambigu."
    )

current_process = "dim_processes_phases"
target_dim_processes_phases = current_catalog + "." + common_schema + "." + current_process
print(target_dim_processes_phases)

all_columns = dim_processes_phases.columns
primary_key = ['id_process_phase']
additional_columns = get_additional_columns(all_columns, primary_key)

if verbose_mode == 'debug':
    display(all_columns)
    print(additional_columns)

ensure_delta_table(dim_processes_phases, target_dim_processes_phases)

handle_table_update(
    dim_processes_phases,
    target_dim_processes_phases,
    primary_key,
    all_columns,
    additional_columns_to_check=additional_columns,
    mode="full"
)
```

## Les 3 nomenclatures TRS



```python
for entity_type in ["measure", "category", "measure_type"]:
    df_entity = trs_nomenclature.filter(F.col("entity_type") == entity_type)

    # La cle porte le cote « un » des deux relations : elle doit etre unique.
    # Le contrat de l'equipe trs_metadata le garantit, on le verifie quand meme.
    doublons = df_entity.groupBy("key").count().filter(F.col("count") > 1)
    if doublons.count() > 0:
        display(doublons)
        raise ValueError(
            f"trs_nomenclature : `key` n'est pas unique pour entity_type = {entity_type}. "
            "Les relations Power BI deviendraient plusieurs-a-plusieurs."
        )

    # Les nomenclatures du projet portent toutes un drapeau `deleted` : on le pose
    # ici pour que cles_nomenclature() reste utilisable sans variante.
    df_entity = df_entity.select("key").withColumn("deleted", F.lit(False))

    current_process = f"dim_trs_{entity_type}"
    target_trs = current_catalog + "." + common_schema + "." + current_process
    print(target_trs)

    all_columns = df_entity.columns
    primary_key = ['key']
    additional_columns = get_additional_columns(all_columns, primary_key)

    ensure_delta_table(df_entity, target_trs)

    handle_table_update(
        df_entity,
        target_trs,
        primary_key,
        all_columns,
        additional_columns_to_check=additional_columns,
        mode="full"
    )
```
