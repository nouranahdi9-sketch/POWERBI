```python
# Jointure dynamique avec dim_parameter : recupere le parameter_id ACTUEL
# pour chaque code. Si l'ID change cote source, il est repris automatiquement
# au prochain run du pipeline, sans modification manuelle necessaire.
# La regle de gestion (aggregation) se base sur le code, qui est la cle
# metier stable -- parameter_id n'est qu'une consequence resolue dynamiquement
# pour permettre la jointure avec fact_batch_measures.
#
# dim_parameter conserve les parametres supprimes (deleted = True) : on ne
# joint que sur les actifs, sinon un code supprime continue de matcher son
# ancien id (filtre cote Power BI) et un code recree a la source sort en double.
dim_parameter_actif = dim_parameter.filter(~F.coalesce(F.col("deleted"), F.lit(False)))

dim_report_parameter_mapping_join = (
    dim_report_parameter_mapping_raw.alias("a")
    .join(
        dim_parameter_actif.alias("b"),
        F.col("a.code") == F.col("b.code"),
        "left"
    )
    .select(
        F.col("b.parameter_id"),
        F.col("a.code"),
        F.col("a.column_label"),
        F.col("a.aggregation"),
        F.col("a.display_order")
    )
)

# Un code qui ne matche aucun parametre actif (ID supprime, code renomme cote
# source, etc.) aurait parameter_id = null. parameter_id est le cote "un" de la
# relation avec fact_batch_measures dans Power BI : une seule valeur vide fait
# echouer le refresh du modele. Ces lignes sont donc ecartees, avec alerte.
codes_non_matches = dim_report_parameter_mapping_join.filter(F.col("parameter_id").isNull())
nb_non_matches = codes_non_matches.count()
if nb_non_matches > 0:
    print(f"ATTENTION : {nb_non_matches} code(s) de dim_report_parameter_mapping ne matchent aucun parameter_id actif dans dim_parameter, lignes ecartees :")
    codes_non_matches.select("code").show(truncate=False)

# Plusieurs parametres actifs pour un meme code : le MERGE sur `code` echouerait
# et parameter_id ne serait plus unique cote Power BI. On bloque explicitement.
codes_doublons = (
    dim_report_parameter_mapping_join
    .filter(F.col("parameter_id").isNotNull())
    .groupBy("code")
    .agg(F.collect_list("parameter_id").alias("parameter_ids"))
    .filter(F.size("parameter_ids") > 1)
)
if codes_doublons.count() > 0:
    codes_doublons.show(truncate=False)
    raise ValueError("dim_report_parameter_mapping : plusieurs parameter_id actifs pour un meme code dans dim_parameter")

dim_report_parameter_mapping = dim_report_parameter_mapping_join.filter(F.col("parameter_id").isNotNull())
```

```python
target_dim_report_parameter_mapping = current_catalog + "." + current_schema + "." + current_process
print(target_dim_report_parameter_mapping)
```

```python
all_columns = dim_report_parameter_mapping.columns
primary_key = ['code']
additional_columns = get_additional_columns(all_columns, primary_key)

if verbose_mode == 'debug':
    display(all_columns)
    print(additional_columns)
```

```python
align_target_schema(
    dim_report_parameter_mapping,
    target_dim_report_parameter_mapping,
    verbose=(verbose_mode == 'debug')
)

if spark.catalog.tableExists(target_dim_report_parameter_mapping):
    # Code renomme cote source (meme parameter_id) : on renomme la ligne cible en
    # place avant le MERGE sur `code`. Sinon le MERGE insererait une nouvelle ligne
    # -- parameter_id en double cote Power BI, et les colonnes propres a la cible
    # (exclure_zero_moyenne, created_at) repartiraient a NULL.
    dim_report_parameter_mapping.createOrReplaceTempView("dim_report_parameter_mapping_source")
    spark.sql(f"""
        MERGE INTO {target_dim_report_parameter_mapping} t
        USING dim_report_parameter_mapping_source s
        ON t.parameter_id = s.parameter_id AND t.code <> s.code
        WHEN MATCHED THEN UPDATE SET t.code = s.code
    """)

    # Le mode update de handle_table_update ne fait jamais de delete : des lignes
    # ecrites avec parameter_id = null par d'anciens runs resteraient en cible et
    # continueraient de bloquer le refresh Power BI.
    spark.sql(f"DELETE FROM {target_dim_report_parameter_mapping} WHERE parameter_id IS NULL")

handle_table_update(
    dim_report_parameter_mapping,
    target_dim_report_parameter_mapping,
    primary_key,
    all_columns,
    additional_columns_to_check=additional_columns,
    mode=execution_mode
)
```
