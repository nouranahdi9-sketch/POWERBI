# main_translations

Orchestrateur du **job des traductions**, autonome et indépendant du pipeline BI.

Il alimente `{translations_schema}` (le catalogue commun) avec `dim_trad_language` et
les 11 tables `dim_trad_*`. Plusieurs projets les consomment ; un seul job les
écrit.

## Planification

À caler sur le rythme d'apparition de **nouvelles clés** (variétés, catégories de
notes, paramètres), pas sur celui des corrections de libellés : entre deux runs,
une clé nouvellement créée s'affiche sans libellé dans les rapports.

Le job dure quelques minutes — un run quotidien réduit ce délai à moins de 24 h
pour un coût négligeable.

## Dépendance avec les pipelines BI

Ces tables doivent exister avant qu'un modèle sémantique ne les lise. Si un
pipeline BI est déployé sur un environnement neuf, lancer ce job en premier.
